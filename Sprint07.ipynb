Función para obtener los datos:
- `while True` para aceptar tantos fallos del usuario como sea necesario antes de dar por bueno el input.
- `try` para gestionar el posible ValueError en caso de no usar un número entero.
- La condición if...break, apra permanecer en el bucle mientras no se introduzca un dígito de una longitud igual a 9.
- Con anys_exp queremos limpiar la entrada para modificar comas por puntos y permanecer en el bucle hasta obtener un dígito viable.


```python
# Es crea la funció per demanar les dades:
def obtenir_dades():
  nom = input("Introdueixi el seu nom i cognoms: ")
  while True:
    try:
      telefon = int(input("Introdueixi el seu número de contacte, si us plau, usi nou dígits: "))
      if len(telefon) == 9:
        break
      else:
        print("Si us plau, usi nou dígits per introduir el telèfon.")
    except ValueError:
      print("Si us plau, usi únicament números.")
    while True:
      try:
        anys_exp_brut = input("Introdueixi els anys d'experiència a l'empresa: ")
        anys_exp_net = anys_exp_brut.replace(",", ".")
        anys_exp = float(anys_exp_net)
        break
      except ValueError:
        print("Si us plau, usi únicament números.")
    while True:
      try:
        con_cat = int(input("Introdueixi el seu domini del català. Usi: 0 sense coneixement, 1 fins a B1, 2 B2 i 3 per C1 o superior: "))
        if con_cat in {0,1,2,3}:
          break
        else:
          print("Si us plau, introdueixi un dels dígits permesos: 0, 1, 2 o 3.")
      except ValueError:
        print("Si us plau, utilitzo els dígits permesos: 0, 1, 2 o 3.")



  dom_exel = input("Introdueixi el seu domini de l'Excel. Usi: 0 per nivell bàsic, 1 ús de funcions, 2 taules dinàmiques i 3 Visual Basic i Macros: ")
  tit_academic = input("Introdueixi la seva titulació acadèmica més alta. Usi: 0 ESO, 1 FP, 2 Universitat i 3 Màster o superior: ")
  return nom, telefon, anys_exp, con_cat, dom_exel, tit_academic
```
