Función para obtener los datos mediante inputs.

````python
def obtenir_dades():
  nom = input("Introdueixi el seu nom i cognoms: ")
  telefon = input("Introdueixi el seu número de contacte, si us plau, usi nou dígits: ")
  anys_exp = input("Introdueixi els anys d'experiència a l'empresa: ")
  con_cat = input("Introdueixi el seu domini del català. Usi: 0 sense coneixement, 1 fins a B1, 2 B2 i 3 per C1 o superior: ")
  dom_exel = input("Introdueixi el seu domini de l'Excel. Usi: 0 per nivell bàsic, 1 ús de funcions, 2 taules dinàmiques i 3 Visual Basic i Macros: ")
  tit_academic = input("Introdueixi la seva titulació acadèmica més alta. Usi: 0 ESO, 1 FP, 2 Universitat i 3 Màster o superior: ")
  return nom, telefon, anys_exp, con_cat, dom_exel, tit_academic
````

La siguiente función limpia los datos (tipo y comas/puntos) y revisa que las condiciones necesarias para el uso de estos datos sean las correctas.
Se usan las siguientes condiciones:

- `while True` para aceptar tantos fallos del usuario como sea necesario antes de dar por bueno el input.
- `try` para gestionar el posible ValueError en caso de no usar un número entero.
- La condición if...break, apra permanecer en el bucle mientras no se introduzca un dígito de una longitud igual a 9.
- Con anys_exp queremos limpiar la entrada para modificar comas por puntos y permanecer en el bucle hasta obtener un dígito viable.
- Para los tres últimos inputs se acepta únicamente números enteros y dentro del conjunto {0,1,2,3}.


````python
def validar_dades(nom, telefon, anys_exp, con_cat, dom_exel, tit_academic):
  if not (telefon.isdigit() and len(telefon) == 9):
    return False, "Error: El telèfon ha de contenir exactament 9 dígits numèrics."
  try:
        anys_net = anys_exp.replace(",", ".")
        anys_exp_final = float(anys_net)
  except ValueError:
        return False, "Error: Els anys d'experiència han de ser un número vàlid."
  try:
      con_cat_final = int(con_cat)
      if con_cat_final not in {0, 1, 2, 3}:
          return False, "Error: El domini del català ha de ser 0, 1, 2 o 3."
  except ValueError:
      return False, "Error: El domini del català ha de ser un número enter."
  try:
      dom_exel_final = int(dom_exel)
      if dom_exel_final not in {0, 1, 2, 3}:
          return False, "Error: El domini de l'Excel ha de ser 0, 1, 2 o 3."
  except ValueError:
      return False, "Error: El domini de l'Excel ha de ser un número enter."
  try:
      tit_academic_final = int(tit_academic)
      if tit_academic_final not in {0, 1, 2, 3}:
          return False, "Error: La titulació acadèmica ha de ser 0, 1, 2 o 3."
  except ValueError:
      return False, "Error: La titulació acadèmica ha de ser un número enter."
  return True, (nom, telefon, anys_exp_final, con_cat_final, dom_exel_final, tit_academic_final)
````

Función para calcular la puntuación obtenida.

````python
def calcular_puntuacio(nom, telefon, anys_exp_final, con_cat_final, dom_exel_final, tit_academic_final):
  anys_exp_calcul = anys_exp_final * 1.5
  puntuacio_final = anys_exp_calcul + con_cat_final + dom_exel_final + tit_academic_final
  return nom, telefon, puntuacio_final
````

Función para categorizar cada perfil candidato. La función se parará con el décimo candidato.

````python
def classificar_candidat(puntuacio_final):
  if puntuacio_final < 8:
    classificacio = "Prioritat baixa"
  elif puntuacio_final < 12:
    classificacio = "Perfil adequat"
  else:
    classificacio = "Prioritat alta"
  return classificacio
````

Función para mostrar un informe del perfil individual del candidato.

````python
def mostrar_individual(nom, telefon, puntuacio_final, classificacio):
  print(f' El candidat {nom} amb telèfon de contacte {telefon} ha obtingut una puntuació de: {puntuacio_final}, quedant classificat com {classificacio}')
````

Función para mostrar un informe de todos los candidatos presentados, ordenados por puntuación.

Para ello se crea un diccionario con la classificación por puntuación deseada como clave y un conjunto de la información de cada candidato como valor.

Se clasifican los candidatos según la prioridad extraída de la lista de candidatos creada en la función `main`con toda la información resultante de las funciones anteriores.

Por último se imprime un resumen con el nombre, teléfono y puntuación de cada candidato, clasificado según el valor `classificació`.

````python
def mostrar_informe(llista_candidats):
  classificacio_perfils = {
      "Prioritat alta": [],
      "Perfil adequat": [],
      "Prioritat baixa": []
  }
  for candidat in llista_candidats:
    classificacio = candidat["classificació"]
    if classificacio in classificacio_perfils:
      classificacio_perfils[classificacio].append(candidat)
  for categoria, perfils in classificacio_perfils.items():
    perfils_ordenats = sorted(perfils, key=lambda candidat:candidat["puntuació"], reverse = True)
    print(categoria)
    for perfil in perfils_ordenats:
      print(f' - {perfil["nom"]}, {perfil["telèfon"]}, puntuació: {perfil["puntuació"]}')
````

Por último se monta una función que enseñe y recorra todos los candidatos.

Esta función se basa en un `while`que se rompe a la que se introduzca la información **válida** del décimo candidato.

El bucle recorre las diferentes funciones hasta crear un diccionario con la información de los candidatos que se van introduciendo.

Una vez se llega al décimo se rompe y se llama la última función, que usando esta lista crea la clasificación final y ordenada.

````python
def main_seleccio():
  llista_candidats = []
  candidats = 0
  while candidats < 10:
    print(f'Candidat {candidats + 1} de 10')
    while True:
      nom, telefon, anys_exp, con_cat, dom_exel, tit_academic = obtenir_dades()
      candidat_valid, info_candidat = validar_dades(nom, telefon, anys_exp, con_cat, dom_exel, tit_academic)
      if candidat_valid:
        nom, telefon, anys_exp_final, con_cat_final, dom_exel_final, tit_academic_final = info_candidat
        break
      else:
        print("Informació no vàlida, cal tornar a indicar-la.")
    nom, telefon, puntuacio_final = calcular_puntuacio(nom, telefon, anys_exp_final, con_cat_final, dom_exel_final, tit_academic_final)
    classificacio = classificar_candidat(puntuacio_final)
    candidat_actual = {
        "nom": nom,
        "telèfon": telefon,
        "puntuació": puntuacio_final,
        "classificació": classificacio
    }
    llista_candidats.append(candidat_actual)
    candidats += 1
    mostrar_individual(nom, telefon, puntuacio_final, classificacio)
  mostrar_informe(llista_candidats)
````

````python
main_seleccio()
````

Cuando se llega al candidato error, la función vuelve a pedir la información para ese número de candidato, sin pasar al siguiente. Una vez se introducen los diez, no pide un undécimo, sino que enseña como estos quedan clasificados y las puntuaciones obtenidas.

Con el método map() que modificará todas las filas de la columna seleccionada según la coincidencia clave:valor del nuevo diccionario indicado.

De esta manera, se crea una nueva columna `factors_comportament`con el factor de comportamiento según el valor indicado en la columna `codi_comportament`.

````python
def obtenir_segment_i_factor(df_p):
  factors_comportament_dicc = {
      0: 0.10,
      1: 0.25,
      2: 0.50,
      3: 0.75,
      4: 0.90
  }
  df_p["factor_comportament"] = df_p["codi_comportament"].map(factors_comportament_dicc)
  return df_p
````

````python
def obtenir_factor_edat(df_p):
  factor_edat_dicc = {
      "18-25": 1.20,
      "26-40": 1.00,
      "41-60": 0.85,
      "60+": 0.70
  }
  df_p["factor_edat"] = df_p["grup_edat"].map(factor_edat_dicc)
  return df_p
````
